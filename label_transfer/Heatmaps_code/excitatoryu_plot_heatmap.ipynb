```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# choose which half to show: "first" or "second"
half = "first"

cols = exc_all.columns.to_list()
mid = len(cols) // 2

if half == "first":
    cols_use = cols[:mid]
else:
    cols_use = cols[mid:]

data = exc_all.loc[:, cols_use]

plt.figure(figsize=(14, 8))
ax = sns.heatmap(
    data,
    vmin=-0.6, vmax=0.6,
    cmap="coolwarm",
    annot=True, fmt=".2f",            # <-- show corr values
    annot_kws={"size": 7},            # <-- make numbers smaller if crowded
    linewidths=0.2
)

# force all x labels to show
ax.set_xticks(np.arange(len(cols_use)) + 0.5)
ax.set_xticklabels(cols_use, rotation=90)
plt.tight_layout()
plt.show()
```
